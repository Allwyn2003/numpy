```python
import numpy as np
```

```python
#Array Creation
project=np.array([[1,2,3,4],[41,42,43,44],[99,98,97,96]])
project
```

```python
#Data Exploration
project.shape
project.size
project.ndim
```

```python
#Data Transformation
project.T
```

```python
#Adding a new column in an existing array
newcol = np.array([[5], [45], [100]])
res = np.hstack((project,newcol))
print(res)
```

```python
#Updation
project[2][1]=100
print(project)
```

```python
#Deletion
delt=np.delete(project,1)
print(delt)
```

```python
#Datatype Handling
project[0][2]=47.4
print(project)
```

```python
#Copy of an arrayc
pro=np.copy(project)
print(pro)
```

```python
#View of an same array
pro.view()
print(pro)
```
